## Step 1 — Start learning PyTorch

Buku menjelaskan bahwa framework yang dibangun pada Chapter 13 paling menyerupai **PyTorch**.

Alasannya:

- sebelumnya kita bekerja dengan NumPy dan framework buatan sendiri;
- setelah memahami apa yang terjadi di bawah hood, penggunaan framework menjadi lebih praktis;
- framework membantu mengurangi bug dan mempercepat runtime;
- pengguna juga dapat mempelajari serta menggunakan code dari komunitas.

Buku secara khusus merekomendasikan PyTorch bagi pembaca yang memiliki latar belakang NumPy karena API-nya terasa familiar.

```python
comparison = {
    "NumPy": "Basic matrix library",
    "Chapter 13 framework": "Framework built from scratch",
    "PyTorch": "Practical deep learning framework"
}

for name, role in comparison.items():
    print(f"{name}: {role}")
```

### Hubungan Chapter 13 → PyTorch

Chapter 13 sengaja membangun framework sederhana sebagai jembatan menuju framework nyata.

Dengan demikian, pembelajaran tidak langsung dimulai dari API tingkat tinggi. Pembaca terlebih dahulu memahami mekanisme seperti tensor, autograd, layers, loss functions, dan optimization.

```python
fig, ax = plt.subplots(figsize=(10, 3.8))
ax.axis("off")

items = [
    ("NumPy", 0.12),
    ("Build a framework", 0.37),
    ("Understand the API", 0.62),
    ("PyTorch", 0.88)
]

for label, x in items:
    ax.text(x, 0.55, label, ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5", fill=False, linewidth=1.2))

for (_, x1), (_, x2) in zip(items[:-1], items[1:]):
    ax.annotate("", xy=(x2-0.055,0.55), xytext=(x1+0.055,0.55),
                arrowprops=dict(arrowstyle="->", linewidth=1.4))

ax.set_title("From low-level understanding to a practical framework")
plt.show()
```

## Step 2 — Start another deep learning course

Buku menekankan bahwa satu buku atau satu course tidak cukup untuk seluruh pendidikan Deep Learning.

Konsep yang sama sebaiknya dipelajari dari beberapa perspektif.

Tujuannya bukan sekadar mengumpulkan course, tetapi membuat fundamental menjadi **second nature** melalui:

**Learn → Relearn → Practice → Become familiar**.

```python
cycle = ["Learn", "Relearn", "Practice", "Internalize"]

fig, ax = plt.subplots(figsize=(8.5, 4.5))
ax.bar(cycle, [1, 2, 3, 4])
ax.set_ylabel("Conceptual progression")
ax.set_title("Repeated exposure to Deep Learning concepts")
ax.set_yticks([])
ax.grid(axis="y", alpha=0.2)
plt.show()
```

**Catatan:** tinggi batang hanya menunjukkan urutan proses secara konseptual, bukan ukuran kemampuan yang terukur.

## Step 3 — Grab a mathy deep learning textbook

Setelah memiliki intuisi Deep Learning, buku menyarankan untuk memperdalam mathematics.

Pendekatan yang dijelaskan adalah:

**Implement → Observe → Reverse-engineer the mathematics**

Dengan pengalaman coding, mathematics yang sebelumnya abstrak dapat dipahami kembali melalui operasi yang benar-benar digunakan dalam model.

Buku menyebut *Deep Learning* oleh Ian Goodfellow, Yoshua Bengio, dan Aaron Courville sebagai contoh textbook tingkat berikutnya pada saat buku ditulis.

```python
topics = ["Algebra", "Calculus", "Statistics"]
emphasis = [1, 1, 1]

fig, ax = plt.subplots(figsize=(7.5, 4.2))
ax.barh(topics, emphasis)
ax.set_xlabel("Core mathematical area")
ax.set_title("Mathematics suggested for deeper study")
ax.set_xticks([])
ax.grid(axis="x", alpha=0.15)
plt.show()
```

## Step 4 — Start a blog, and teach deep learning

Buku memberikan penekanan kuat pada teaching.

Menulis tutorial memaksa kita untuk:

- memahami konsep;
- memecah konsep menjadi bagian kecil;

## Step 5 — Twitter

Buku menyarankan mengikuti percakapan AI untuk tetap **up-to-date**.

Pendekatan yang dijelaskan:

1. Temukan AI researchers yang dikagumi.
2. Follow mereka.
3. Ikuti orang-orang yang mereka follow.
4. Gunakan feed tersebut untuk menemukan paper, ideas, dan diskusi baru.

Tujuannya bukan sekadar social media activity, tetapi menjadi bagian dari conversation di bidang AI.

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.axis("off")

nodes = [
    ("Researchers", 0.18, 0.55),
    ("People they\nfollow", 0.50, 0.75),
    ("AI conversation", 0.82, 0.55),
    ("Papers & ideas", 0.50, 0.30)
]

for label, x, y in nodes:
    ax.text(x, y, label, ha="center", va="center",
            bbox=dict(boxstyle="round,pad=0.5", fill=False))

arrows = [
    ((0.27,0.58),(0.42,0.70)),
    ((0.58,0.70),(0.73,0.58)),
    ((0.73,0.51),(0.58,0.35)),
    ((0.42,0.35),(0.27,0.51)),
]
for start, end in arrows:
    ax.annotate("", xy=end, xytext=start,
                arrowprops=dict(arrowstyle="->", linewidth=1.2))

ax.set_title("Building awareness of the AI research conversation")
plt.show()
```

## Step 7 — Acquire access to a GPU

Buku menekankan hubungan antara **faster experimentation** dan **faster learning**.

Menurut source, GPU dapat memberikan training time yang jauh lebih cepat, dengan klaim sekitar **10–100× faster training times** pada konteks yang dibahas buku.

Contoh akses yang disebut buku termasuk GPU NVIDIA dan free K80s yang pada saat penulisan dapat digunakan melalui Google Colab.

```python
speedup_range = [10, 100]

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(["Lower end", "Upper end"], speedup_range)
ax.set_ylabel("Training speed multiplier (×)")
ax.set_title("GPU speedup range reported by the book")
ax.grid(axis="y", alpha=0.2)
for i, value in enumerate(speedup_range):
    ax.text(i, value + 2, f"{value}×", ha="center")
plt.show()
```

**Interpretasi:** angka 10–100× adalah rentang yang disebut buku, bukan jaminan bahwa setiap model akan memperoleh speedup tersebut.
