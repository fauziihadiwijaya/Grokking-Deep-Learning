## 2. Theoretical deep-dive

### 2.1 What is deep learning?
Buku menjelaskan deep learning sebagai bagian dari machine learning yang terutama menggunakan *artificial neural networks* (jaringan saraf tiruan), yang secara sederhana terinspirasi oleh cara kerja otak manusia. Machine learning sendiri merupakan bagian dari bidang *artificial intelligence* (AI).

### 2.2 Why learn it?
Trask membahas penerapan deep learning pada *computer vision*, pengenalan suara, *machine translation*, serta berbagai tugas yang membutuhkan kemampuan mengenali pola. Chapter ini juga menyinggung kemungkinan otomatisasi pekerjaan yang memerlukan keahlian. Pembahasan mengenai dampak terhadap pekerjaan merupakan pandangan penulis dan sebaiknya dipahami sebagai kemungkinan, bukan kepastian.

### 2.3 Principles versus frameworks
Framework seperti Torch, TensorFlow, dan Keras membantu pengguna membangun model. Namun, memahami prinsip dasar di baliknya juga penting agar kita mengerti proses yang dijalankan framework dan tidak hanya bergantung pada satu API atau alat tertentu.

### 2.4 Intuition-first mathematics
Penulis berusaha menjelaskan matematika yang diperlukan melalui aritmetika, aljabar, dan analogi yang mudah dipahami. Tujuannya bukan menghilangkan matematika, melainkan membantu pembaca memahami alasan dan mekanisme di balik perhitungan.

### 2.5 Project-based learning
Buku menekankan pembelajaran melalui praktik: menghubungkan konsep dengan kode, mengamati hasilnya, lalu memahami kapan dan mengapa suatu metode digunakan.

## 3. AI, Machine Learning, and Deep Learning
Diagram lingkaran bertingkat berikut menunjukkan hubungan cakupan antara AI, machine learning, dan deep learning. Ukuran lingkaran hanya ilustrasi, bukan proporsi berdasarkan data.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

fig, ax = plt.subplots(figsize=(8, 5.5))
ax.add_patch(Circle((0.48, 0.48), 0.43, color="#DCEBFA", ec="#275D8C", lw=2))
ax.add_patch(Circle((0.48, 0.45), 0.29, color="#B9D9F2", ec="#275D8C", lw=2))
ax.add_patch(Circle((0.48, 0.40), 0.15, color="#F4C76B", ec="#8A5B00", lw=2))
ax.text(0.48, 0.82, "Artificial Intelligence (AI)", ha="center", fontsize=13, weight="bold")
ax.text(0.48, 0.60, "Machine Learning", ha="center", fontsize=12, weight="bold")
ax.text(0.48, 0.40, "Deep\nLearning", ha="center", va="center", fontsize=11, weight="bold")
ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_aspect("equal"); ax.axis("off")
ax.set_title("Hubungan konsep (tidak berdasarkan skala)", fontsize=14, pad=12)
plt.show()
```

**Interpretation:** Deep learning merupakan salah satu pendekatan dalam machine learning, sedangkan machine learning termasuk dalam bidang AI. Diagram ini tidak berarti bahwa semua sistem AI menggunakan neural network.

## 4. Learning loop used throughout the book
Chapter 1 memperkenalkan pendekatan praktis yang digunakan dalam buku. Pola “membuat prediksi, membandingkan hasil, lalu memperbarui model” diperkenalkan secara eksplisit pada Chapter 2 dan dibahas lebih lanjut pada chapter berikutnya. Diagram ini berfungsi sebagai gambaran awal.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

fig, ax = plt.subplots(figsize=(10, 2.8))
labels = [("PREDICT", "Gunakan input dan model saat ini"), ("COMPARE", "Bandingkan prediksi dengan target"), ("LEARN", "Sesuaikan model untuk mengurangi error")]
xs = [0.04, 0.37, 0.70]
for x, (title, desc) in zip(xs, labels):
    box = FancyBboxPatch((x, 0.28), 0.26, 0.48, boxstyle="round,pad=0.02,rounding_size=0.03",
                         facecolor="#EAF2F8", edgecolor="#356A8A", linewidth=1.5)
    ax.add_patch(box)
    ax.text(x+0.13, 0.59, title, ha="center", va="center", weight="bold", fontsize=12)
    ax.text(x+0.13, 0.42, desc, ha="center", va="center", fontsize=8, wrap=True)
for x1, x2 in [(0.30,0.37),(0.63,0.70)]:
    ax.add_patch(FancyArrowPatch((x1,0.52),(x2,0.52),arrowstyle="-|>",mutation_scale=15,color="#356A8A",lw=1.5))
ax.text(0.5, 0.08, "Gambaran awal: mekanisme pembelajaran dibahas lebih lanjut pada chapter berikutnya.",
        ha="center", fontsize=9, style="italic")
ax.set_xlim(0,1); ax.set_ylim(0,1); ax.axis("off")
plt.show()
```
